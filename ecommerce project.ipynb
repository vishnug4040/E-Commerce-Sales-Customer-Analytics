```python
sales_products = sales.merge(product,on="product_id",how="left")
sales_products.head()
```

```python
sales_products["total_cost"] = (sales_products["quantity"] * sales_products["cost_price"])
```

```python
sales_products["profit"] = (sales_products["line_total"] -sales_products["total_cost"])
```

```python
sales_products["profit_margin"] = (sales_products["profit"] /
sales_products["line_total"]) * 100
```

```python
product_sales = (sales_products
    .groupby("product_name")["line_total"]
    .sum()
    .sort_values(ascending=False))
product_sales
```

```python
state_distribution = customer["state"].value_counts()

print(state_distribution)
```

```python
state_distribution.plot(kind="bar", figsize=(10, 5))

plt.title("Customer Distribution by State")
plt.xlabel("State")
plt.ylabel("Number of Customers")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
bins = [0, 18, 25, 35, 45, 60, 100]
labels = ["Under 18", "18–25", "26–35",
          "36–45", "46–60", "Above 60"]

customer["age_group"] = pd.cut(
    customer["age"],
    bins=bins,
    labels=labels,
    include_lowest=True
)

age_distribution = customer["age_group"].value_counts().sort_index()

print(age_distribution)
```

```python
age_distribution.plot(kind="bar", figsize=(8, 5))

plt.title("Customer Distribution by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Number of Customers")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
orders_customers = orders.merge(customer,
on="customer_id",
how="left")
```

```python
sales_customer = sales.merge(orders_customers,
    on="order_id",
    how="left")
```

```python
state_sales = (sales_customer
    .groupby("state")["line_total"]
    .sum()
    .sort_values(ascending=False))
```

```python
plt.figure(figsize=(12,6))

state_sales.plot(kind="barh")

plt.title("Sales by State")
plt.xlabel("State")
plt.ylabel("Sales Amount")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
orders_by_state = (orders_customers
    .groupby("state")["order_id"]
    .nunique()
    .sort_values(ascending=False))

print(orders_by_state)
```

```python
orders_by_state.head(8).plot( kind="pie",figsize=(8,8),autopct="%1.1f%%")

plt.title("Orders by State")
plt.ylabel("")
plt.show()
```

```python
sales_orders = sales.merge(orders[["order_id", "order_date"]],
    on="order_id",
    how="left")

sales_orders["month"] = (
    sales_orders["order_date"]
    .dt.to_period("M")
    .astype(str))

monthly_sales = (sales_orders
    .groupby("month")["line_total"]
    .sum())
monthly_sales
```

```python
plt.figure(figsize=(14,6))

monthly_sales.plot(kind="line", marker="o")

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
sales_orders["year"] = sales_orders["order_date"].dt.year

yearly_sales = (sales_orders.groupby("year")["line_total"].sum())

print(yearly_sales)
yearly_sales.plot(
    kind="bar",
    figsize=(8,5))

plt.title("Sales by Year")
plt.xlabel("Year")
plt.ylabel("Sales")
plt.show()
print("Highest Sales Year:",yearly_sales.idxmax())
```

```python
carrier_delivery = (shipping
    .groupby("carrier")["actual_days"]
    .mean()
    .sort_values())
print(carrier_delivery)
```

```python
product_rating = (reviews.merge(
        product[["product_id", "product_name"]],
        on="product_id",
        how="left")
    .groupby("product_name")["rating"]
    .mean()
    .sort_values(ascending=False))
print(product_rating.head(10))
```
